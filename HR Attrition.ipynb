```python
import pandas as pd
df = pd.read_csv('HR-Employee-Attrition.csv')
df.head(20)
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace('_','')
df.columns
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicates:", df.duplicated().sum())
```

```python
df["attrition_flag"] = (df["attrition"] == "Yes").astype(int)
print(df.head(10))
```

```python
import pandas as pd
df["age_group"] = pd.cut(df["age"], bins=[17, 25, 35, 45, 55, 100],
                         labels=["18-25", "26-35", "36-45", "46-55", "56+"])
print(df[["age","age_group"]].head(10))
```

```python
df["income_band"] = pd.cut(df["monthlyincome"], bins=[0, 3000, 6000, 10000, 25000],
                        labels=["Low", "Medium", "High", "Very High"])
print(df[["monthlyincome", "income_band"]].head(20))
```

```python
df["tenure_band"] = pd.cut(df["yearsatcompany"], bins=[-1, 2, 5, 10, 50],
                           labels=["0-2 yrs", "3-5 yrs", "6-10 yrs", "10+ yrs"])
print(df[["yearsatcompany","tenure_band"]].head(10))
```

```python
print(df[["attrition", "attrition_flag", "age_group", "income_band", "tenure_band"]].head())
print("Attrition rate %:", round(df["attrition_flag"].mean() * 100, 2))
```

```python
df.to_csv("HR_clean.csv", index=False)
print("Saved HR_clean.csv", df.shape)
```
